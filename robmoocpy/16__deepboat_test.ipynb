````python

def make_phi_hat():
    model_path = 'phi_hat_model.keras'
    data_path = 'deepboatdata.txt'

    if os.path.exists(model_path):
        return tf.keras.models.load_model(model_path)
    
    else:
        if os.path.exists(data_path):
            print("Training new model...")
            
            phi_hat_model = tf.keras.Sequential([
                tf.keras.layers.Input(shape=(5,)),
                tf.keras.layers.Dense(16, activation='relu'),
                tf.keras.layers.Dense(16, activation='relu'),
                tf.keras.layers.Dense(3)
            ])

            phi_hat_model.compile(optimizer='adam', loss='mse')
            exp_data = np.loadtxt('deepboatdata.txt')
            X_train = exp_data[:, :5]
            y_train = exp_data[:, 5:]
            phi_hat_model.fit(X_train, y_train, epochs=5, batch_size=32)

            # we should save the generated model to use it later
            phi_hat_model.save(model_path)
            return phi_hat_model
    
        elif not os.path.exists('deepboatdata.txt'):
            print("Error: deepboatdata.txt not found.")
            return None
````

````python
phi_hat_model = make_phi_hat()
````

````python
phi_hat_model_2 = make_phi_hat()
phi_hat_model_2
````

````python

def phi_hat(vx,vy,w,u1,u2, phi_hat_model):
    input_data = np.array([[vx, vy, w, u1, u2]])
    dvx, dvy, dw = phi_hat_model.predict(input_data)[0]
    # sortie = tolist(phi_hat_model.predict(input_data).flatten())
    return dvx, dvy, dw
````

Use a horizontal concatenation (axis=1). For example, if `p` is `2×m` so that `p.T` is `m×2`, and you have another matrix `Q` of shape `m×k` you want on the right:

```python
C = np.concatenate((p.T, Q), axis=1)
# or equivalently
C = np.hstack((p.T, Q))
```

If you want to add a single column vector `v` (length `m`) on the right of `p.T`:

```python
v = np.arange(m)          # shape (m,)
C = np.hstack((p.T, v[:, None]))  # v[:, None] -> shape (m,1)
```
